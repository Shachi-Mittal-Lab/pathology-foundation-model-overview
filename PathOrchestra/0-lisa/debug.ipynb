```python
import numpy as np
from pathlib import Path
from collections import Counter

data_root = Path(r"E:\[PROJ]_DCIS\DATA\training_patches\annotation_scheme-0\as0_all_tiles")
files = sorted(data_root.glob("*.npz"))

all_pixels = Counter()

for f in files:
    data = np.load(f)
    mask = data['label']
    vals, counts = np.unique(mask, return_counts=True)
    for v, c in zip(vals, counts):
        all_pixels[int(v)] += int(c)

total = sum(all_pixels.values())
print(f"Total files: {len(files)}")
print(f"Total pixels: {total:,}")
print(f"\nClass distribution:")
for k, v in sorted(all_pixels.items()):
    print(f"  class {k:4d}: {v:12,} pixels  ({100*v/total:.2f}%)")
```

```python
import numpy as np
from pathlib import Path
from collections import Counter

data_root = Path(r"E:\[PROJ]_DCIS\DATA\training_patches\annotation_scheme-0\as0_tissue-tiles")
files = sorted(data_root.glob("*.npz"))

all_values = Counter()
files_with_unexpected = []

for f in files:
    data = np.load(f)
    mask = data['label']
    unique_vals = np.unique(mask)
    
    # flag any file with values outside expected set {0, 1, 2, 3, 255}
    expected = {0, 1, 2, 3, 255}
    unexpected = set(unique_vals.tolist()) - expected
    if unexpected:
        files_with_unexpected.append((f.name, unexpected))
    
    for v, c in zip(*np.unique(mask, return_counts=True)):
        all_values[int(v)] += int(c)

print("=== GLOBAL CLASS DISTRIBUTION ===")
total = sum(all_values.values())
for k, v in sorted(all_values.items()):
    print(f"  class {k:4d}: {v:12,} pixels  ({100*v/total:.2f}%)")

print(f"\n=== UNEXPECTED VALUES ===")
if files_with_unexpected:
    for fname, vals in files_with_unexpected[:10]:
        print(f"  {fname}: {vals}")
    print(f"  ... {len(files_with_unexpected)} files total with unexpected values")
else:
    print("  None — all values are in {0, 1, 2, 3, 255}")
```

```python
import numpy as np

# your pixel counts
counts = np.array([
     22_603_945,   # class 0 - background
     17_106_765,   # class 1 - epithelium
     53_085_776,   # class 2 - stroma
      6_489_088,   # class 3 - other
])

# inverse frequency weighting
weights = 1.0 / counts
weights = weights / weights.sum()  # normalize to sum to 1
weights = weights * len(counts)    # scale so mean weight = 1

print("Class weights:")
for i, w in enumerate(weights):
    print(f"  class {i}: {w:.4f}")
```
