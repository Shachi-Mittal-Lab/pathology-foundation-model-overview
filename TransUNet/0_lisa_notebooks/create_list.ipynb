```python
from pathlib import Path
import random, shutil

src = Path(r"E:\PROJ_DCIS\training_patches\annotation_scheme-0\TEMP_all-patches")
dst = Path(r"E:\PROJ_DCIS\training_patches\annotation_scheme-0\TEMP_half-patches")
N = 7000  # how many to move

dst.mkdir(parents=True, exist_ok=True)

# pick file types you consider "patches"
files = [p for p in src.iterdir() if p.is_file() and p.suffix.lower() in {".npz", ".npy", ".png", ".tif", ".tiff"}]

if N > len(files):
    raise ValueError(f"Requested {N} files, but only found {len(files)} in {src}")

for p in random.sample(files, N):
    shutil.move(str(p), str(dst / p.name))

print(f"Moved {N} files from {src} -> {dst}")
```
